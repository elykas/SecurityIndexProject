```python
import pandas as pd
```

```python
flights = pd.read_csv("../data/flightsData.csv",parse_dates=['CHSTOL'])
flights.head()
```

```python
flights = flights.copy()
```

```python
flights.columns = (
    flights.columns
        .str.strip()
        .str.lower()
        .str.replace(' ', '_')
        .str.replace('"', '')
)
```

```python
column_rename_map = {
        "choper": "airline_code",
        "chfltn": "flight_number",
        "choperd": "airline_name",
        "chstol": "scheduled_takeoff",
        "chptol": "actual_takeoff",
        "chaord": "terminal_direction",
        "chloc1": "destination_airport_code",
        "chloc1d": "destination_airport_name_eng",
        "chloc1th": "destination_city_name_heb",
        "chloc1t": "destination_city_name_eng",
        "chloc1ch": "destination_country_name_heb",
        "chlocct": "destination_country_name_eng",
        "chterm": "terminal",
        "chcint": "international_indicator",
        "chckzn": "security_zone",
        "chrmine": "status_eng",
        "chrminh": "status_heb"
    }

flights.rename(columns=column_rename_map, inplace=True)
flights.head()
```

```python
flights['date'] = flights['scheduled_takeoff'].dt.date
flights.head()
```

```python
flights['status_eng'] = flights['status_eng'].str.lower()
```

```python
flights['canceled'] = flights['status_eng'] == 'canceled'
```

```python
daily_cancel_rate = flights.groupby('date')['canceled'].mean().reset_index()
```

```python
daily_cancel_rate.rename(columns={'canceled': 'cancel_rate'}, inplace=True)
```

```python
daily_cancel_rate['cancel_rate'] *= 100
```
