```python
import pandas as pd
```

```python
trains = pd.read_csv("../data/trainData.csv", sep='|')
trains.head()

```

```python
trains = trains.copy()
```

```python
trains['date'] = pd.to_datetime(trains['shana'].astype(str) + '-' + trains['hodesh'].astype(str), format='%Y-%m')
trains['date'].head()
```

```python
grouped = trains.groupby('date')['ahuz_bitzua'].mean().reset_index()
grouped.set_index('date', inplace=True)
grouped.head(15)
```
