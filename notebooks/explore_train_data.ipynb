```python
import pandas as pd
```

```python
trains = pd.read_csv("../data/trainsData.csv", sep='|')
trains.head()
```

```python
trains = trains.copy()
```

```python
trains.columns = (
        trains.columns
        .str.strip()
        .str.lower()
        .str.replace(' ', '_')
        .str.replace('"', '')
    )
```

```python
column_rename_map = {
    "shana": "year",
    "hodesh": "month",
    "train_trip_cd": "trip_id",
    "first_train_station_nm": "origin_station",
    "last_train_station_nm": "destination_station",
    "rishui_all": "total_registered",
    "rishui_only": "registered_only",
    "bitzua_only": "executed_only",
    "ahuz_bitzua": "execution_rate"
    }

trains.rename(columns=column_rename_map, inplace=True)
```

```python
trains = trains.copy()
```

```python
trains['date'] = pd.to_datetime(trains['year'].astype(str) + '-' + trains['month'].astype(str), format='%Y-%m')
```

```python
monthly_execution_rate = trains.groupby('date')['execution_rate'].mean().reset_index()
```
